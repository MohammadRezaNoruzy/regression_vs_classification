```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.linear_model import LinearRegression

# regression
np.random.seed(42)
X_class = np.random.randn(100, 2)
y_class = (X_class[:, 0] + X_class[:, 1] > 0).astype(int)

X_reg = np.random.rand(100, 1) * 10
y_reg = 2 * X_reg[:, 0] + np.random.randn(100) * 2

#  classification
x1 = np.linspace(-2, 2, 100)
x2 = -x1

#  regression line
w = np.polyfit(X_reg[:, 0], y_reg, 1)
y_reg_pred = np.polyval(w, X_reg[:, 0])

# 2D  classification
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.scatter(X_class[y_class == 0, 0], X_class[y_class == 0, 1], color='red', label='Class 0')
plt.scatter(X_class[y_class == 1, 0], X_class[y_class == 1, 1], color='blue', label='Class 1')
plt.plot(x1, x2, color='black', linestyle='--', label='Separating Line')
plt.title('2D Classification: Separating Hyperplane')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid()

# 2D  regression
plt.subplot(1, 2, 2)
plt.scatter(X_reg, y_reg, color='blue', label='Data Points')
plt.plot(X_reg, y_reg_pred, color='red', label='Fitted Line')
plt.title('2D Regression: Fitted Hyperplane')
plt.xlabel('Feature')
plt.ylabel('Target')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()

# 3D  classification
fig = plt.figure(figsize=(12, 6))
ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(X_class[:, 0], X_class[:, 1], y_class, c=y_class, cmap='coolwarm', s=50, edgecolor='k')
ax1.set_title('3D Classification: Separating Hyperplane')
ax1.set_xlabel('Feature 1')
ax1.set_ylabel('Feature 2')
ax1.set_zlabel('Class Label')

# 3Dregression
ax2 = fig.add_subplot(122, projection='3d')
ax2.scatter(X_reg[:, 0], y_reg, np.zeros_like(y_reg), c=y_reg, cmap='viridis', s=50, edgecolor='k')
ax2.plot(X_reg[:, 0], y_reg_pred, np.zeros_like(y_reg_pred), color='red', label='Fitted Line')
ax2.set_title('3D Regression: Fitted Hyperplane')
ax2.set_xlabel('Feature')
ax2.set_ylabel('Target')
ax2.set_zlabel('Value')
model_reg = LinearRegression()
model_reg.fit(X_reg, y_reg)

# 3D regression plot
x1_range = np.linspace(X_reg[:, 0].min(), X_reg[:, 0].max(), 20)
y_pred_plane = model_reg.predict(x1_range.reshape(-1, 1))
x1_range_grid, y_pred_plane_grid = np.meshgrid(x1_range, y_pred_plane)
ax2.plot_surface(x1_range_grid, y_pred_plane_grid, np.zeros_like(y_pred_plane_grid), alpha=0.5, cmap=plt.cm.coolwarm)

plt.tight_layout()
plt.show()
```
